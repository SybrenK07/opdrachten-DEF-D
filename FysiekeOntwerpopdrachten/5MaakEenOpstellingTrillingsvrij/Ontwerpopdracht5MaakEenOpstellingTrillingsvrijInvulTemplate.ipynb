## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
f = 5 #Hz
w = 2*np.pi * f

# we weten dat A_m / A_v gelijk moet zijn aan 1/3 voor een verkleining van een factor van 3. Dus moet 1 / (1 - (w^2)/(w_0^2))
# ook gelijk zijn anan 1/3. Omdat het in beide gevallen gaat om een absoluut geldt, w^2/w_0^2 moet gelijk zijn aan 4 dus moet
# w_0^2 gelijk zijn aan een kwart van w^2 minimaal.

w_0 = np.sqrt(w**2 / 4)
f_0 = w_0 / (2 * np.pi)

print(w_0 / np.pi, 'pi')
print(f_0)
```

```python
# Voeg hier je plot in.

def ampoverdracht(f_0):
    return np.abs(1 / (1 - (f_0**2 / 2.5**2)))

f_0 = np.append(np.linspace(0.001,10,100000),np.linspace(10,1000,1000))
amp = ampoverdracht(f_0)

plt.figure()
plt.plot(f_0,amp,'k.')
plt.xscale("log")
plt.yscale("log")
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
# dx = g / w_0^2 en met de minimale w_0^2 die we eerder hebben berekend kunnen we het door rekenen

g = 9.81

dx = g / (w_0**2)
print(dx)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

def C(m, dx):
    return (m * g) / dx

m = np.linspace(0,1,100000) # kg
C_1 = C(m, dx)
C_2 = C(m, 2*dx)
C_5 = C(m, 5*dx)

plt.figure()
plt.plot(m,C_1,'k--',label="Berekende dx")
plt.plot(m,C_2,'b--',label="Berekende dx keer 2")
plt.plot(m,C_5,'r--',label="Berekende dx keer 5")
```

## Opdracht 7: Plotten maar

```python
# Voeg hier je plots in.
f = np.array([1.3,3.0,4.8,7.1,9.3])

A_m = np.array([0.51,0.7,0.69,0.66,0.82])
A_v = np.array([0.15,0.85,3.14,5.30,4.74])

A_ratio = A_m / A_v

f_err = np.array([1,0.3,0.3,0.4,0.3])

A_m_err = np.array([0.8,0.2,0.1,0.3,0.3])
A_v_err = np.array([0.8,0.3,0.2,0.4,0.5])
A_err = np.sqrt((A_m_err / A_m)**2 + (A_v_err / A_v)**2) * A_ratio

plt.figure()
plt.plot(f_0,ampoverdracht(f_0,2 * np.pi * 2.5**2),'b.')
plt.plot(f_0,ampoverdracht(f_0,9.81/5.5),'r.')
plt.errorbar(f,A_ratio,f_err,A_err,'k.', label="Metingen")
plt.xscale("log")
plt.yscale("log")
plt.xlim(0.2,50)
plt.ylim(0.05, 10)
plt.show()
```
